```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("CIA 1.csv")
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df['Asking_Price_INR'].dropna(), kde=True, color='skyblue', bins=30)
plt.title('Distribution of Asking Price (INR)')
plt.xlabel('Asking Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='Category', palette='Set2')
plt.title('Count of Devices by Category')
plt.xlabel('Device Category')
plt.ylabel('Count')
plt.show()
```

```python
sns.boxplot(data=df, x='Condition', y='Final_Price_INR', palette='pastel')
plt.title('Final Price vs Condition (Outlier Detection)')
plt.xlabel('Device Condition')
plt.ylabel('Final Price (INR)')

plt.show()
```
